```python
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
import torch

model_id = "google/gemma-3-1b-it"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    attn_implementation="eager"
)

def generate_text(prompt, max_new_tokens=150):
    chat = [{"role": "user", "content": prompt}]
    inputs = tokenizer.apply_chat_template(chat, tokenize=False, add_generation_prompt=True)
    input_ids = tokenizer.encode(inputs, add_special_tokens=False, return_tensors="pt").to(model.device)
    outputs = model.generate(input_ids=input_ids, max_new_tokens=max_new_tokens)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)


generate_text("Translate from English to Italian (just the sentence)\nbanana", 128)
```

```python
text_file = "./dataset/ita.txt"

import io
import random
from sklearn.model_selection import train_test_split
from datasets import Dataset

SAMPLES = 10_000
def create_dataset(file_path, test_size=0.1, val_size=0.111):
    inputs = []
    outputs = []

    with io.open(file_path, encoding='UTF-8') as f:
        lines = f.read().strip().split('\n')

    for line in lines[:SAMPLES]:
        parts = line.split('\t')
        if len(parts) >= 2:
            inputs.append(parts[0])
            outputs.append(parts[1])

    X_train, X_test, y_train, y_test = train_test_split(inputs, outputs, test_size=test_size, random_state=42)

    X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=val_size, random_state=42)

    train_dataset = Dataset.from_dict({"input": X_train, "output": y_train})
    val_dataset = Dataset.from_dict({"input": X_val, "output": y_val})
    test_dataset = Dataset.from_dict({"input": X_test, "output": y_test})

    return train_dataset, val_dataset, test_dataset

train_dataset, val_dataset, test_dataset = create_dataset(text_file)

print("Train dataset size:", len(train_dataset))
print("Validation dataset size:", len(val_dataset))
print("Test dataset size:", len(test_dataset))
```

```python
import evaluate
from tqdm import tqdm

def evaluate_bleu(model_generate_func, test_dataset):
    bleu_metric = evaluate.load("bleu")

    print("\nGenerating translations for the test set and computing BLEU score...")
    for item in tqdm(test_dataset):
        input_text = item["input"]
        reference_text = item["output"]

        prompt = f"Return only the Italian translation of this sentence: {input_text}"

        generated_text_raw = generate_text(prompt, max_new_tokens=150)

        if "model\n" in generated_text_raw:
            hypothesis_text = generated_text_raw.split("model\n", 1)[1].strip()
        else:
            hypothesis_text = generated_text_raw.strip() # Fallback

        bleu_metric.add(prediction=hypothesis_text, reference=[reference_text])

    results = bleu_metric.compute()

    return results["bleu"]

bleu_score = evaluate_bleu(generate_text, test_dataset)
print(f"\nBLEU Score on the test set: {bleu_score:.4f}")
```

```python
print("\n--- Sample Translations from Test Set ---")
num_samples_to_print = 10 # You can adjust how many samples to print
random_indices = random.sample(range(len(test_dataset)), min(num_samples_to_print, len(test_dataset)))

for i, idx in enumerate(random_indices):
    sample_item = test_dataset[idx]
    input_text = sample_item["input"]
    expected_output = sample_item["output"]

    prompt = f"Return only the Italian translation of this sentence: {input_text}"
    generated_text_raw = generate_text(prompt, max_new_tokens=150)

    if "model\n" in generated_text_raw:
        translated_output = generated_text_raw.split("model\n", 1)[1].strip()
    else:
        translated_output = generated_text_raw.strip()
    bleu_metric = evaluate.load("bleu")
    bleu_metric.add(prediction=translated_output, reference=[expected_output])
    bleu_score = bleu_metric.compute()["bleu"]

    print(f"Sample {i+1}:")
    print(f"  Input: {input_text}")
    print(f"  Expected: {expected_output}")
    print(f"  Translated: {translated_output}")
    print(f"  BLEU Score: {bleu_score}")
    print("-" * 30)
```

```python
def create_prompt_and_tokenize(examples):
    all_input_ids = []
    all_labels = []

    for i in range(len(examples["input"])):
        prompt_text = f"Return only the Italian translation of this sentence: {examples['input'][i]}"
        response_text = f"{examples['output'][i]}"

        prompt_tokenized = tokenizer(prompt_text, add_special_tokens=False)
        response_tokenized = tokenizer(response_text, add_special_tokens=False)

        prompt_len = len(prompt_tokenized['input_ids'])

        full_input_ids = prompt_tokenized['input_ids'] + response_tokenized['input_ids']

        full_input_ids.append(tokenizer.eos_token_id)

        labels = list(full_input_ids)

        for j in range(prompt_len):
            labels[j] = -100

        max_length = 128

        if len(full_input_ids) < max_length:
            pad_len = max_length - len(full_input_ids)
            full_input_ids.extend([tokenizer.pad_token_id] * pad_len)
            labels.extend([-100] * pad_len) # Mask padding in labels too

        elif len(full_input_ids) > max_length:
            full_input_ids = full_input_ids[:max_length]
            labels = labels[:max_length]

        all_input_ids.append(full_input_ids)
        all_labels.append(labels)

    return {"input_ids": all_input_ids, "labels": all_labels}


tokenized_train_dataset = train_dataset.map(create_prompt_and_tokenize, batched=True, batch_size=64, remove_columns=train_dataset.column_names)
tokenized_val_dataset = val_dataset.map(create_prompt_and_tokenize, batched=True, batch_size=64, remove_columns=val_dataset.column_names)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("--- Data Correctly Prepared for Fine-Tuning ---")
print("Sample of tokenized training data:")
print(tokenized_train_dataset[0])
```

```python
from transformers import TrainingArguments, Trainer
import numpy as np
import evaluate
from peft import LoraConfig, get_peft_model
import torch

output_dir = "./gemma_finetuned_checkpoints"

lora_config = LoraConfig(
    r=8,
    target_modules=["q_proj", "o_proj", "k_proj", "v_proj", "gate_proj", "up_proj", "down_proj"],
    bias="none",
    task_type="SEQ_2_SEQ_LM",
)

peft_model = get_peft_model(model, lora_config)

peft_model.print_trainable_parameters()

training_args = TrainingArguments(
    output_dir=output_dir,
    num_train_epochs=3,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    learning_rate=2e-5,
    weight_decay=0.01,
    eval_strategy="epoch",
    logging_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    report_to="none",
    gradient_accumulation_steps=4,
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={'use_reentrant': False},
    dataloader_num_workers=0,
    dataloader_pin_memory=False,
)

trainer = Trainer(
    model=peft_model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_val_dataset,
    tokenizer=tokenizer,
)

print("--- Trainer Configured with PEFT ---")
```

```python
print("Starting the fine-tuning process...")

trainer.train()

print("--- Fine-Tuning Complete ---")
print(f"The best model checkpoint is saved in the directory: {training_args.output_dir}")
```

```python
import evaluate
import random
from tqdm import tqdm

def evaluate_and_show_samples(model_generate_func, test_dataset, num_samples=10):
    bleu_metric = evaluate.load("bleu")
    all_predictions = []
    all_references = []

    print("--- Evaluating Test Set and Generating Samples ---")
    for item in tqdm(test_dataset, desc="Translating test set"):
        prompt = f"Return only the Italian translation of this sentence: {item['input']}"
        generated_text_raw = model_generate_func(prompt, max_new_tokens=150)

        if "model\n" in generated_text_raw:
            hypothesis_text = generated_text_raw.split("model\n", 1)[1].strip()
        else:
            hypothesis_text = generated_text_raw.strip()  # Fallback

        all_predictions.append(hypothesis_text)
        all_references.append([item["output"]])

    overall_bleu_score = bleu_metric.compute(predictions=all_predictions, references=all_references)["bleu"]
    print(f"\nOverall BLEU Score on the test set: {overall_bleu_score:.4f}")
    print("-" * 50)

    print(f"\n--- Displaying {num_samples} Random Samples ---\n")
    random_indices = random.sample(range(len(test_dataset)), min(num_samples, len(test_dataset)))

    for i, idx in enumerate(random_indices):
        input_text = test_dataset[idx]["input"]
        expected_output = all_references[idx][0]
        translated_output = all_predictions[idx]

        sample_bleu_score = evaluate.load("bleu").compute(predictions=[translated_output], references=[[expected_output]])["bleu"]

        print(f"Sample {i+1}:")
        print(f"  Input: {input_text}")
        print(f"  Expected: {expected_output}")
        print(f"  Translated: {translated_output}")
        print(f"  Sample BLEU Score: {sample_bleu_score:.4f}")
        print("-" * 30)

# Execute the function with the previously defined model and test set
evaluate_and_show_samples(generate_text, test_dataset)
```
